### Formatting the files to be accepted by sql

```python
import pandas as pd
from pathlib import Path


accounts_df = pd.read_csv('ACCOUNTS.csv')
credit_score_df = pd.read_csv('CUSTOMER_CREDIT_SCORE.csv')
transactions_df = pd.read_csv('TRANSACTIONS.csv')
demographics_df = pd.read_csv('CUSTOMER_DEMOGRAPHICS.csv')

print("Files loaded Successfully")
```

```python
def describe_data_types():
    print("1. Customer Demographics Table:")
    display(demographics_df.info())
    
    print("\n2. Customer Accounts Table:")
    display(accounts_df.info())

    print("\n3. Customer Credit Score Table:")
    display(credit_score_df.info())

    print("\n4. Customer Transactions Table:")
    display(transactions_df.info())

    

describe_data_types()
```

```python
def inspect_tables():
    print("1. Customer Demographics Table")
    display(demographics_df.head(10))

    print("\n2. Customer Account Table")
    display(accounts_df.head(10))

    print("\n3. Customer Credit Score Table:")
    display(credit_score_df.head(10))

    print("\n4. Customer Transactions Table:")
    display(transactions_df.head(10))


inspect_tables()
```

```python
def check_null():
    print("1. Customer Demographics Table - Null Counts")
    display(demographics_df.isna().sum())

    print("\n2. Customer Account Table - Null Counts")
    display(accounts_df.isna().sum())

    print("\n3. Customer Credit Score Table - Null Counts")
    display(credit_score_df.isna().sum())

    print("\n4. Customer Transactions Table - Null Counts")
    display(transactions_df.isna().sum())

check_null()
```

```python
# Changing dates to datetime format
def changing_date_formats():
    demographics_df['DOB'] = pd.to_datetime(demographics_df['DOB'], format='mixed', errors='coerce')
    accounts_df['OPEN_DATE'] = pd.to_datetime(accounts_df['OPEN_DATE'], format= 'mixed', errors='coerce')
    credit_score_df['LAST_UPDATED'] = pd.to_datetime(credit_score_df['LAST_UPDATED'], format= 'mixed', errors='coerce')
    transactions_df['TXN_DATE'] = pd.to_datetime(transactions_df['TXN_DATE'], format= 'mixed', errors='coerce')

changing_date_formats()
print("Date formats changed successfully")
```

```python
describe_data_types()
```

```python
inspect_tables()
```

```python
transactions_df['AMOUNT'] = transactions_df['AMOUNT'].astype(str).str.replace(r"[R,\s]", "", regex=True).replace({"nan": None, "None": None, "": None})
print("Successfully replaced inconsistant values")
```

```python
transactions_df['AMOUNT'] = pd.to_numeric(transactions_df['AMOUNT'])
print("Successfully changed data type")
```

```python
# Creating Silver Tables

silver_accounts_df = accounts_df
silver_credit_score_df = credit_score_df
silver_transactions_df = transactions_df
silver_demographics_df = demographics_df

print("Silver Tables created successfully")
```

```python
# Describing Silver Tables

def describe_data_types():
    print("1. Customer Demographics Silver Table:")
    display(silver_demographics_df.info())
    
    print("\n2. Customer Accounts Silver Table:")
    display(silver_accounts_df.info())

    print("\n3. Customer Credit Score Silver Table:")
    display(silver_credit_score_df.info())

    print("\n4. Customer Transactions Silver Table:")
    display(silver_transactions_df.info())

    

describe_data_types()
```

### Describing Pathing in both os and Path

```python
# Loading silver tables
CURR_DIR = Path.cwd() # Current Working Directory

CSV_FOLDER = CURR_DIR/'Updated_files'
CSV_FOLDER.mkdir(parents=True, exist_ok=True) # creating the target directory

#loading inside the created directory
silver_demographics_df.to_csv(CSV_FOLDER/'demographics.csv', index=False)
silver_accounts_df.to_csv(CSV_FOLDER/'accounts.csv', index=False)
silver_credit_score_df.to_csv(CSV_FOLDER /'credit_score.csv', index = False)
silver_transactions_df.to_csv(CSV_FOLDER/'transactions.csv', index=False)

print(f"Silver tables loaded successfully to {CSV_FOLDER}")
```
